find inference csv files and copy to another folder

```python
import os
import shutil
```

```python
# 定义主目录
data_dir = './data'

# 遍历主目录下的所有子目录和文件
for root, dirs, files in os.walk(data_dir):
    for file in files:
        # 检查文件名是否包含 'inference' 且以 '.csv' 结尾
        if 'inference' in file and file.endswith('.csv'):
            file_path = os.path.join(root, file)  # 获取文件的完整路径
            destination = os.path.join(data_dir, file)  # 目标路径是 ./data 目录下
            
            # 复制文件到 ./data 目录下
            try:
                shutil.copy(file_path, destination)
                print(f"Copied: {file_path} to {destination}")
            except Exception as e:
                print(f"Error copying {file_path}: {e}")
```
